```python
def get_selectivity_metrics(file_path):
    # Step 1: Preprocess the data
    reorganized_data, variable_list = preprocess_data(file_path, normalize=True)

    # Step 2: Initialize output lists
    raw_selectivity = []
    Argmax_list_raw = []
    Argmin_list_raw = []

    # Step 3: Compute selectivity for raw data
    selectivity_list_raw = []
    for animal, subdict in reorganized_data.items():
        for cell, value in subdict.items():
            neuron_data = subdict[cell]
            neuron_activity = neuron_data[:, 0, :]  # Extract activity
            neuron_activity_trial_averaged = np.mean(neuron_activity, axis=1)  # Average over trials
            selectivity_list_raw.append(neuron_activity_trial_averaged)

    # Compute Vinje selectivity for raw data
    selectivity_matrix_raw = np.vstack(selectivity_list_raw)
    for cell in selectivity_matrix_raw:
        raw_selectivity.append(Vinje2000(cell, norm='min_max'))
        Argmax_list_raw.append(np.argmax(cell))
        Argmin_list_raw.append(np.argmin(cell))

    # Step 4: Compute selectivity for variable-subtracted conditions
    selectivity_velocity = compute_condition_selectivity(reorganized_data, variable_list, ["Velocity"])
    selectivity_vl = compute_condition_selectivity(reorganized_data, variable_list, ["Velocity", "Licks"])
    selectivity_vr = compute_condition_selectivity(reorganized_data, variable_list, ["Velocity", "Reward_loc"])
    selectivity_vrl = compute_condition_selectivity(reorganized_data, variable_list, ["Velocity", "Reward_loc", "Licks"])

    return raw_selectivity, selectivity_velocity, selectivity_vl, selectivity_vr, selectivity_vrl


def compute_condition_selectivity(reorganized_data, variable_list, variables_to_remove):
    """
    Helper function to compute selectivity for a given condition by removing specified variables.
    """
    avg_residuals, _, _ = compute_variable_subtracted_residuals(reorganized_data, variable_list, variables_to_remove, quintile=None)
    _, _, _, selectivity = get_min_maxed_residuals_argmin_argmax_selectivity(avg_residuals)
    return selectivity

def compute_condition_argmax(reorganized_data, variable_list, variables_to_remove):
    """
    Helper function to compute selectivity for a given condition by removing specified variables.
    """
    avg_residuals, _, _ = compute_variable_subtracted_residuals(reorganized_data, variable_list, variables_to_remove, quintile=None)
    _, argmax, _, _ = get_min_maxed_residuals_argmin_argmax_selectivity(avg_residuals)
    return selectivity

def compute_condition_argmin(reorganized_data, variable_list, variables_to_remove):
    """
    Helper function to compute selectivity for a given condition by removing specified variables.
    """
    avg_residuals, _, _ = compute_variable_subtracted_residuals(reorganized_data, variable_list, variables_to_remove, quintile=None)
    _, _, argmin, _ = get_min_maxed_residuals_argmin_argmax_selectivity(avg_residuals)
    return argmin

```

```python
def compute_condition_residuals(reorganized_data, variable_list, variables_to_remove):
    """
    Helper function to compute selectivity for a given condition by removing specified variables.
    """
    _, _, residual_activity = compute_variable_subtracted_residuals(reorganized_data, variable_list, variables_to_remove, quintile=None)
    _, _, argmin, _ = get_min_maxed_residuals_argmin_argmax_selectivity(avg_residuals)
    return argmin
```

```python
def get_peak_loc_metrics(file_path):
    # Step 1: Preprocess the data
    reorganized_data, variable_list = preprocess_data(file_path, normalize=True)

    # Step 2: Initialize output lists
    raw_argmax = []

    # Step 3: Compute argmax for raw data
    argmax_list_raw = []
    for animal, subdict in reorganized_data.items():
        for cell, value in subdict.items():
            neuron_data = subdict[cell]
            neuron_activity = neuron_data[:, 0, :]  # Extract activity
            neuron_activity_trial_averaged = np.mean(neuron_activity, axis=1)  # Average over trials
            argmax_list_raw.append(neuron_activity_trial_averaged)

    # Compute Vinje argmax for raw data
    argmax_matrix_raw = np.vstack(argmax_list_raw)
    for cell in argmax_matrix_raw:
        raw_argmax.append(np.argmax(cell))
 
    # Step 4: Compute argmax for variable-subtracted conditions
    argmax_velocity = compute_condition_argmax(reorganized_data, variable_list, ["Velocity"])
    argmax_vl = compute_condition_argmax(reorganized_data, variable_list, ["Velocity", "Licks"])
    argmax_vr = compute_condition_argmax(reorganized_data, variable_list, ["Velocity", "Reward_loc"])
    argmax_vrl = compute_condition_argmax(reorganized_data, variable_list, ["Velocity", "Reward_loc", "Licks"])

    return raw_argmax, argmax_velocity, argmax_vl, argmax_vr, argmax_vrl

```

```python


def t_test_graph(x, y, x_labels, y_title):

    t_stat, p_value_raw_raw = ttest_rel(x, y)

    mean_y = np.mean(y)
    mean_x = np.mean(x)

    sem_y = np.std(y) / np.sqrt(len(y))
    sem_x = np.std(x) / np.sqrt(len(x))

    plt.figure(figsize=(8, 6))

    # Scatter points
    x_positions = [0.4, 0.6]

    mean_offset = 0.02
    plt.errorbar(
        x_positions[0] - mean_offset, mean_y, yerr=sem_y, fmt='o', color='red' 
    )
    plt.errorbar(
        x_positions[1] + mean_offset, mean_x, yerr=sem_x, fmt='o', color='red'

    )

    for i in range(len(y)):
        plt.plot(x_positions, [x[i], y[i]], color='gray', alpha=0.5)

    plt.scatter([x_positions[0]] * len(x), x, color='blue')
    plt.scatter([x_positions[1]] * len(y), y, color='orange')

    # Add labels and title
    plt.xticks(x_positions, x_labels)
    plt.ylabel(y_title)

    # Display p-value with a red line
    y_min = max(max(x), max(y))
    plt.plot([x_positions[0], x_positions[1]], [y_min + y_min/20, y_min + y_min/20], color='red', linewidth=2)
    plt.grid()         
    plt.text(0.5, y_min + y_min/17, f"P-value: {p_value_raw_raw:.3f}", ha='center', color='red')
    
    
y_title = 'Vinje 2000 Selectivity Index'
x_labels_SST_vel = ["Raw Selectivity SST", "Velocity-Subtracted Selectivity SST"]
t_test_graph(selectivity_SST_velocity, SST_raw_selectivity, x_labels_SST_vel, y_title)


y_title = 'Vinje 2000 Selectivity Index'
x_labels_SST_vl = ["Raw Selectivity SST", "Velocity and Licks-Subtracted Selectivity SST"]
t_test_graph(selectivity_SST_vl, SST_raw_selectivity, x_labels_SST_vl, y_title)


y_title = 'Vinje 2000 Selectivity Index'
x_labels_SST_vr = ["Raw Selectivity SST", "Velocity and Reward-Subtracted Selectivity SST"]
t_test_graph(selectivity_SST_vr, SST_raw_selectivity, x_labels_SST_vr, y_title)

y_title = 'Vinje 2000 Selectivity Index'
x_labels_SST_vrl = ["Raw Selectivity SST", "Velocity,Lick,Reward-Subtracted Selectivity SST"]
t_test_graph(selectivity_SST_vrl, SST_raw_selectivity, x_labels_SST_vrl, y_title)


# groups_SST = [SST_raw_selectivity, selectivity_SST_velocity, selectivity_SST_vl, selectivity_SST_vr, selectivity_SST_vrl]
# group_labels_SST = ["Raw Selectivity", "Velocity-Subtracted Selectivity", "Velocity and Licks-Subtracted Selectivity", "Velocity and Reward-Subtracted Selectivity","Velocity,Lick,Reward-Subtracted Selectivity"]


```

```python
def get_amrgmin_argmax_difference(raw, y):
    bin_edges = np.arange(0, 51, 5)
    bin_centers = bin_edges[:-1] + 2.5
    bin_labels = [f"{start}-{start + 4}" for start in bin_edges[:-1]]

    raw_hist, _ = np.histogram(raw, bins=bin_edges)
    y_hist, _ = np.histogram(y, bins=bin_edges)

    difference = (y_hist - raw_hist)

    z_list = []
    
    for value in difference:
        value = value / len(raw)
        z_list.append(value)
        
    return z_list
```

```python
def return_min_max_quintile_residuals(residual_activity, raw=False):

    first_quintile_argmin = []
    last_quintile_argmin = []
    
    first_quintile_argmax = []
    last_quintile_argmax = []

    for animal, subdict in residual_activity.items():
        for cell, value in subdict.items():
            residual_data = residual_activity[animal][cell]
            if raw:
                residual_data = residual_data[:, 0, :]
            quintiles = split_into_quintiles(residual_data)
            first_quintile = quintiles[0]
            last_quintile = quintiles[-1]
            argmin_loc = np.mean(first_quintile, axis=1)
            argmin_loc = np.argmin(argmin_loc)
            first_quintile_argmin.append(argmin_loc)
            argmin_loc = np.mean(last_quintile, axis=1)
            argmin_loc = np.argmin(argmin_loc)
            last_quintile_argmin.append(argmin_loc)
            argmax_loc = np.mean(first_quintile, axis=1)
            argmax_loc = np.argmax(argmax_loc)
            first_quintile_argmax.append(argmax_loc)
            argmax_loc = np.mean(last_quintile, axis=1)
            argmax_loc = np.argmax(argmax_loc)
            last_quintile_argmax.append(argmax_loc)
            
    return first_quintile_argmin, last_quintile_argmin, first_quintile_argmax, last_quintile_argmax




```

```python
def compare_and_plot_quintiles_grid(data_dict):
    """
    Compare the first and last quintiles within each list in the dictionary
    and plot the results as subplots in a 5x2 grid (argmin and argmax per row).
    
    Parameters:
    - data_dict: dict, keys are descriptive names of the datasets,
                 and values are lists containing two datasets to compare.
    """
    num_rows = len(data_dict) // 2  # Assuming 2 keys (argmin, argmax) per condition
    fig, axs = plt.subplots(num_rows, 2, figsize=(15, 5 * num_rows))
    
    # Flatten axs if num_rows == 1
    if num_rows == 1:
        axs = np.array([axs])

    # Create a mapping from condition to row index
    condition_to_row = {}
    row = 0
    for key in sorted(data_dict.keys()):
        base_condition = key.replace('argmin_', '').replace('argmax_', '')
        if base_condition not in condition_to_row:
            condition_to_row[base_condition] = row
            row += 1

    for idx, (condition, datasets) in enumerate(data_dict.items()):
        # Determine column index based on 'argmin' or 'argmax'
        if "argmin" in condition:
            col_idx = 0  # Left column for argmin
        elif "argmax" in condition:
            col_idx = 1  # Right column for argmax
        else:
            continue

        # Get the base condition to find the correct row
        base_condition = condition.replace('argmin_', '').replace('argmax_', '')
        row_idx = condition_to_row[base_condition]

        # Unpack first and last quintile data
        data1, data2 = datasets

        # Select the appropriate subplot
        ax = axs[row_idx, col_idx]
        ax.set_title(condition)
        
        # Plot the comparison
        t_test_graph(
            data1, data2,
            x_labels=["First Quintile", "Last Quintile"],
            y_title="Distance From Reward (cm)",
            ax=ax
        )

    plt.tight_layout()
    plt.show()


    
    
compare_and_plot_quintiles_grid(data_dict_SST)
compare_and_plot_quintiles_grid(data_dict_NDNF)
compare_and_plot_quintiles_grid(data_dict_EC)
```

```python
def t_test_graph(x, y, x_labels, y_title, ax=None):
    t_stat, p_value_raw_raw = ttest_rel(x, y)

    mean_y = np.mean(y)
    mean_x = np.mean(x)

    sem_y = np.std(y) / np.sqrt(len(y))
    sem_x = np.std(x) / np.sqrt(len(x))

    if ax is None:
        ax = plt.gca()  # Get current axes

    # Scatter points
    x_positions = [0.4, 0.6]
    mean_offset = 0.02

    ax.errorbar(
        x_positions[0] - mean_offset, mean_y, yerr=sem_y, fmt='o', color='red' 
    )
    ax.errorbar(
        x_positions[1] + mean_offset, mean_x, yerr=sem_x, fmt='o', color='red'
    )

    for i in range(len(y)):
        ax.plot(x_positions, [x[i], y[i]], color='gray', alpha=0.5)

    ax.scatter([x_positions[0]] * len(x), x, color='blue')
    ax.scatter([x_positions[1]] * len(y), y, color='orange')

    # Add labels and title
    ax.set_xticks(x_positions)
    ax.set_xticklabels(x_labels)
    ax.set_ylabel(y_title)

    # Display p-value with a red line
    y_max = max(max(x), max(y))
    ax.plot([x_positions[0], x_positions[1]], [y_max + y_max/20, y_max + y_max/20], color='red', linewidth=2)
    ax.grid()         
    ax.text(0.5, y_max + y_max/17, f"P-value: {p_value_raw_raw:.3f}", ha='center', color='red')
```

```python
def get_trial_averaged_raw_data(reorganized_data_group):
    
    trial_av_group_raw = []

    for animal, subdict in reorganized_data_group.items():
        for cell, value in subdict.items():
            neuron_data = reorganized_data_group[animal][cell]
            neuron_data = neuron_data[:, 0, :]
            trial_averaged = np.mean(neuron_data, axis=1)
            trial_av_group_raw.append(trial_averaged)

    trial_av_group_raw = np.vstack(trial_av_group_raw)   
    
    return trial_av_group_raw

trial_averaged_raw_SST = get_trial_averaged_raw_data(reorganized_data_SST)
trial_averaged_raw_NDNF = get_trial_averaged_raw_data(reorganized_data_NDNF)
trial_averaged_raw_EC = get_trial_averaged_raw_data(reorganized_data_EC)
```

```python
def get_min_maxed_residuals_argmin_argmax_selectivity(avg_residuals):
    argmax_list = []
    argmin_list = []
    positive_selectivity_list = []
    negative_selectivity_list = []
    normalized_average_residuals_min_max = []
    
    for cell in avg_residuals:
        positive_normalized_cell = []
        negative_normalized_cell = []
        for bins in cell:
            norm_bin = (bins - np.min(cell)) / (np.max(cell) - np.min(cell))
            positive_normalized_cell.append(norm_bin)
            
            norm_bin_negative = np.absolute(1-norm_bin)
            negative_normalized_cell.append(norm_bin_negative)
        
        positive_array = np.array(positive_normalized_cell)
        positive_selectivity = Vinje2000(positive_array, norm='None')
        positive_selectivity_list.append(positive_selectivity)
        
        negative_array = np.array(negative_normalized_cell)
        negative_selectivity = Vinje2000(negative_array, norm='None')
        negative_selectivity_list.append(negative_selectivity)
        normalized_average_residuals_min_max.append(positive_normalized_cell)
        argmax_list.append(np.argmax(cell))
        argmin_list.append(np.argmin(cell))

    avg_residuals_min_max = np.vstack(normalized_average_residuals_min_max)
    positive_selectivity = np.vstack(positive_selectivity_list)
    negative_selectivity = np.vstack(negative_selectivity_list)

    return avg_residuals_min_max, argmax_list, argmin_list, positive_selectivity, negative_selectivity
```

```python
def plot_cumulative_selectivity(selectivity_history_dict):
    cumulative_selectivity = []
    n_bins = 100
    cdf_prob_bins = np.arange(1., n_bins + 1.) / n_bins
    for model_seed in selectivity_history_dict:
        selectivity = selectivity_history_dict[model_seed]['Output']

        nonresponsive_idx = np.where(selectivity==0)[0]
        selectivity[nonresponsive_idx] = 128 #penalize units that are silent across all patterns

        selectivity =1 - selectivity / 128  # convert selectivity to fraction active patterns (per unit)

        selectivity = np.sort(selectivity[:])
        quantiles = [np.quantile(selectivity, pi) for pi in cdf_prob_bins]
        cumulative_selectivity.append(quantiles)

    cumulative_selectivity = np.array(cumulative_selectivity)
    mean_selectivity = np.mean(cumulative_selectivity, axis=0)

    SD = np.std(cumulative_selectivity, axis=0)
    SEM = SD / np.sqrt(cumulative_selectivity.shape[0])

    return cumulative_selectivity.flatten(), mean_selectivity, cdf_prob_bins, SD
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import sem

def get_mean_sem_by_animal(split_animal_IDs_list, split_selectivity_list):
       
    num_parts = len(split_animal_IDs_list)
    mean_values = []
    sem_values = []

    for i in range(num_parts):
        split_animal_IDs = np.array(split_animal_IDs_list[i]) 
        split_values = np.array(split_selectivity_list[i])   
      
        unique_animals = np.unique(split_animal_IDs)
        
        animal_means = []
        
        for animal in unique_animals:
            animal_mask = split_animal_IDs == animal
            animal_values = split_values[animal_mask]
            animal_means.append(np.mean(animal_values))  
        
        mean_values.append(np.mean(animal_means))
        sem_values.append(sem(animal_means))

    return mean_values, sem_values


mean_values_SST_raw_positive, sem_values_SST_raw_positive = get_mean_sem_by_animal(split_animal_IDs_list_SST_raw_positive, split_selectivity_list_SST_raw_positive)
mean_values_NDNF_raw_positive, sem_values_NDNF_raw_positive = get_mean_sem_by_animal(split_animal_IDs_list_NDNF_raw_positive, split_selectivity_list_NDNF_raw_positive)
mean_values_EC_raw_positive, sem_values_EC_raw_positive = get_mean_sem_by_animal(split_animal_IDs_list_EC_raw_positive, split_selectivity_list_EC_raw_positive)

mean_values_SST_raw_negative, sem_values_SST_raw_negative = get_mean_sem_by_animal(split_animal_IDs_list_SST_raw_negative, split_selectivity_list_SST_raw_negative)
mean_values_NDNF_raw_negative, sem_values_NDNF_raw_negative = get_mean_sem_by_animal(split_animal_IDs_list_NDNF_raw_negative, split_selectivity_list_NDNF_raw_negative)
mean_values_EC_raw_negative, sem_values_EC_raw_negative = get_mean_sem_by_animal(split_animal_IDs_list_EC_raw_negative, split_selectivity_list_EC_raw_negative)

mean_values_SST_velocity_positive, sem_values_SST_velocity_positive = get_mean_sem_by_animal(split_animal_IDs_list_SST_velocity_positive, split_selectivity_list_SST_velocity_positive)
mean_values_NDNF_velocity_positive, sem_values_NDNF_velocity_positive = get_mean_sem_by_animal(split_animal_IDs_list_NDNF_velocity_positive, split_selectivity_list_NDNF_velocity_positive)
mean_values_EC_velocity_positive, sem_values_EC_velocity_positive = get_mean_sem_by_animal(split_animal_IDs_list_EC_velocity_positive, split_selectivity_list_EC_velocity_positive)

mean_values_SST_velocity_negative, sem_values_SST_velocity_negative = get_mean_sem_by_animal(split_animal_IDs_list_SST_velocity_negative, split_selectivity_list_SST_velocity_negative)
mean_values_NDNF_velocity_negative, sem_values_NDNF_velocity_negative = get_mean_sem_by_animal(split_animal_IDs_list_NDNF_velocity_negative, split_selectivity_list_NDNF_velocity_negative)
mean_values_EC_velocity_negative, sem_values_EC_velocity_negative = get_mean_sem_by_animal(split_animal_IDs_list_EC_velocity_negative, split_selectivity_list_EC_velocity_negative)

mean_value_list_positive = [mean_values_SST_raw_positive, mean_values_NDNF_raw_positive, mean_values_EC_raw_positive, mean_values_SST_velocity_positive, mean_values_NDNF_velocity_positive, mean_values_EC_velocity_positive]
mean_value_list_negative = [mean_values_SST_raw_negative, mean_values_NDNF_raw_negative, mean_values_EC_raw_negative, mean_values_SST_velocity_negative, mean_values_NDNF_velocity_negative, mean_values_EC_velocity_negative]

sem_value_list_positive = [sem_values_SST_raw_positive, sem_values_NDNF_raw_positive, sem_values_EC_raw_positive, sem_values_SST_velocity_positive, sem_values_NDNF_velocity_positive, sem_values_EC_velocity_positive]
sem_value_list_negative = [sem_values_SST_raw_negative, sem_values_NDNF_raw_negative, sem_values_EC_raw_negative, sem_values_SST_velocity_negative, sem_values_NDNF_velocity_negative, sem_values_EC_velocity_negative]


def plot_mean_with_horizontal_sem_by_animal(mean_value_list, sem_value_list, title, y_title):
    
    bin_centers = np.arange(1, 11)
    
    plt.figure(figsize=(10, 6))
    
    plt.errorbar(mean_value_list[0], bin_centers, xerr=sem_value_list[0], fmt='o-', color='black', ecolor='black', capsize=8, label="SST GCaMP") 
    plt.errorbar(mean_value_list[1], bin_centers, xerr=sem_value_list[1], fmt='o-', color='red', ecolor='red', capsize=8, label= "NDNF GCaMP")
    plt.errorbar(mean_value_list[2], bin_centers, xerr=sem_value_list[2], fmt='o-', color='gray', ecolor='gray', capsize=8, label= "EC GCaMP")
   
    plt.errorbar(mean_value_list[3], bin_centers, xerr=sem_value_list[3], fmt='o-', color='blue', ecolor='blue', capsize=8, label="SST Velocity-Subtracted") 
    plt.errorbar(mean_value_list[4], bin_centers, xerr=sem_value_list[4], fmt='o-', color='orange', ecolor='orange', capsize=8, label="NDNF Velocity-Subtracted")
    plt.errorbar(mean_value_list[5], bin_centers, xerr=sem_value_list[5], fmt='o-', color='green', ecolor='green', capsize=8, label="EC Velocity-Subtracted")

    plt.ylabel("Percentile of Data")  
    plt.yticks = np.linspace(1,10,10)
    plt.xlabel(f" Mean {y_title}")  
    plt.title(title)
    plt.legend()

    plt.show()


plot_mean_with_horizontal_sem_by_animal(mean_value_list_positive, sem_value_list_positive, "Mean Argmax Per Animal", "Position Bin of Maximum Firing")
plot_mean_with_horizontal_sem_by_animal(mean_value_list_negative, sem_value_list_negative, "Mean Argmin Per Animal", "Position Bin of Minimum Firing")

```
